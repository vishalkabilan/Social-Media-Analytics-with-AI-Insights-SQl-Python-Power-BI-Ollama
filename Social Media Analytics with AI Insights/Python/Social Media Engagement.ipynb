```python
df['Total_Engagement'] = df['Likes'] + df['Comments'] + df['Shares'] + df['Downloads']
```

```python
df.groupby('Platform')['Total_Engagement'].mean().sort_values(ascending=False)
```

```python
df.groupby('Region')['Total_Engagement'].mean()
```

```python
sns.barplot(x='Platform', y='Total_Engagement', data=df)

plt.xticks(rotation=45)

plt.show()
```

```python
sns.scatterplot(x='Followers', y='Total_Engagement', data=df)

plt.show()
```

```python
sns.heatmap(df[['Likes','Comments','Shares','Downloads','Followers','Total_Engagement']].corr(),annot=True)

plt.show()
```
